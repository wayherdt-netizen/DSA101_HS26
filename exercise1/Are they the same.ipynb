```python
%matplotlib inline

import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from pathlib import Path
```

# Do we think the same thing is happening in all of these countries?

This is a hypothetical dataset representing pesticide residues above a safety threshold vs. percentage of organic farms in different countries. How can we tell if trends are consistent across countries?

```python
files = sorted(Path(".").glob("*.csv"))
print(len(files))
print([f.name for f in files][:3])
```

```python
for file in Path('.').glob("*.csv"):
    df = pd.read_csv(file)
    x, y = df['x'], df['y']
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    coeff = np.polyfit(x, y, deg=1)
    corr = np.corrcoef(x, y)[0, 1]
    
    print(f"x_mean={x_mean:.3f}, y_mean={y_mean:.3f}, linear regression fr(x)={coeff[1]:.1f} {coeff[0]:.2f} x, r = {corr:.3f} ")
```

```python
for file in sorted(Path(".").glob("*.csv")):
    df = pd.read_csv(file)

    plt.figure(figsize=(5, 4))
    plt.scatter(df["x"], df["y"])
    plt.title(file.stem)
    plt.xlabel("Organic farms (%)")
    plt.ylabel("Pesticide residues above threshold (%)")
    plt.show()

#Do we think the same thing is happening in all of these countries?
#No. Although the datasets have almost identical means, correlation coefficients and linear regression equations, their scatterplots can reveal very different patterns.
#Therefore, summary statistics alone are insufficient to conclude that the same relationship exists in all countries.
```
